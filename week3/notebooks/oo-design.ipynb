```python
def add_to_class(Class):  #@save
    """Register functions as methods in created class."""
    def wrapper(obj):
        setattr(Class, obj.__name__, obj)
    return wrapper
```

```python
class A:
    def __init__(self):
        self.b = 1

a = A()
print(a.b)
```

```python
@add_to_class(A)
def print_b(self):
    print(self.b)

a.print_b()
```
